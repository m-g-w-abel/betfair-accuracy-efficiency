```python
import datetime
from urllib.request import urlretrieve
from urllib.error import HTTPError
from pathlib import Path
import time
```

```python
BASE_ADDR = "https://promo.betfair.com/betfairsp/prices/"
```

```python
start_date = datetime.date(2008, 9, 2)
end_date = datetime.date.today()
```

```python
current_date = start_date - datetime.timedelta(days=-1)
while current_date <= end_date:
    filename = f"dwbfpricesukwin{current_date.day:02}{current_date.month:02}{current_date.year}.csv"
    p = Path("data/spcsv/" + filename)
    if p.exists():
        print(f"{filename} already downloaded - skipping")
    else:
        try:
            urlretrieve(BASE_ADDR + filename, "data/spcsv/" + filename)
            print(f"{filename} downloaded - success")
        except HTTPError as e:
            if e.code == 403:
                print(f"{filename} ratelimited - stalling")
                time.sleep(2)
                continue
        except:
            continue

    current_date += datetime.timedelta(days=1)
    
```
